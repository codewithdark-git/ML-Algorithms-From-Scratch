## 1. Vectors

Vectors are the fundamental building blocks of ML - they represent data points, features, weights, and gradients.

```python
import numpy as np
import matplotlib.pyplot as plt

# Vector creation
v = np.array([3, 4])
w = np.array([1, 2])

print(f"v = {v}")
print(f"w = {w}")

# Vector addition
print(f"\nv + w = {v + w}")
print(f"v - w = {v - w}")
print(f"2 * v = {2 * v}")

# Dot product (inner product)
dot = np.dot(v, w)
print(f"\nv \u00b7 w = {dot}")
print(f"v @ w = {v @ w}")

# Geometric interpretation: v · w = ||v|| ||w|| cos(θ)
norm_v = np.linalg.norm(v)
norm_w = np.linalg.norm(w)
cos_theta = dot / (norm_v * norm_w)
theta = np.arccos(cos_theta)
print(f"||v|| = {norm_v:.4f}")
print(f"||w|| = {norm_w:.4f}")
print(f"cos(θ) = {cos_theta:.4f}")
print(f"θ = {np.degrees(theta):.2f}°")

# Orthogonality check
u = np.array([1, 0])
v_orth = np.array([0, 1])
print(f"\nu · v_orth = {u @ v_orth} (orthogonal)")
```
