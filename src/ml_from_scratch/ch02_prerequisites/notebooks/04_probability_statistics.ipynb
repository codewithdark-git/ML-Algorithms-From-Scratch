## 1. Probability Fundamentals

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import itertools

# Probability axioms:
# 1. P(A) ≥ 0 for any event A
# 2. P(Ω) = 1 (sample space)
# 3. P(A ∪ B) = P(A) + P(B) for disjoint A, B

# Discrete probability example: Coin flip
coin_probs = {'H': 0.5, 'T': 0.5}
print(f"Coin: {coin_probs}")
print(f"Sum: {sum(coin_probs.values())}")

# Joint probability
joint = {('H', 'H'): 0.25, ('H', 'T'): 0.25,
         ('T', 'H'): 0.25, ('T', 'T'): 0.25}
print(f"\nTwo coin flips: {joint}")

# Marginal probability
p_H_first = joint[('H', 'H')] + joint[('H', 'T')]
p_H_second = joint[('H', 'H')] + joint[('T', 'H')]
print(f"P(H first) = {p_H_first}")
print(f"P(H second) = {p_H_second}")

# Conditional probability: P(A|B) = P(A,B) / P(B)
p_H1_given_H2 = joint[('H', 'H')] / p_H_second
print(f"P(H1|H2) = {p_H1_given_H2}")

# Independence: P(A,B) = P(A)P(B)
indep = all(abs(joint[k] - coin_probs[k[0]] * coin_probs[k[1]]) < 1e-10
          for k in joint)
print(f"Independent: {indep}")
```
