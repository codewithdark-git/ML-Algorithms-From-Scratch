## 1. NumPy Arrays vs Python Lists

NumPy arrays are homogeneous, fixed-size, and support vectorized operations.

```python
import numpy as np
import sys

# Python list
py_list = [1, 2, 3, 4, 5]
print(f"Python list: {py_list}")
print(f"List size: {sys.getsizeof(py_list)} bytes")

# NumPy array
np_array = np.array([1, 2, 3, 4, 5])
print(f"NumPy array: {np_array}")
print(f"Array size: {np_array.nbytes} bytes (data only)")
print(f"Array dtype: {np_array.dtype}")
print(f"Array shape: {np_array.shape}")
print(f"Array ndim: {np_array.ndim}")

# Performance comparison
import time

size = 1_000_000
py_list_large = list(range(size))
np_array_large = np.arange(size)

# Python list - loop
start = time.perf_counter()
result_py = [x * 2 for x in py_list_large]
t_py = time.perf_counter() - start

# NumPy - vectorized
start = time.perf_counter()
result_np = np_array_large * 2
t_np = time.perf_counter() - start

print(f"\nList comprehension: {t_py:.4f}s")
print(f"NumPy vectorized: {t_np:.4f}s")
print(f"Speedup: {t_py/t_np:.1f}x")
```
